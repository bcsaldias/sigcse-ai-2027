```python
# ── Helpers ───────────────────────────────────────────────────────────────────
def matches(label, keywords):
    return any(kw in label.lower() for kw in keywords)

def hex_to_rgba(hex_color, alpha=0.35):
    h = hex_color.lstrip("#")
    r, g, b = int(h[0:2],16), int(h[2:4],16), int(h[4:6],16)
    return f"rgba({r},{g},{b},{alpha})"

def col_y(n):
    if n == 1: return [0.5]
    return [0.02 + i * 0.96 / (n - 1) for i in range(n)]

def flow_proportional_y(items, flows_dict, pad_norm=8/600):
    flows = [flows_dict.get(item, 0) for item in items]
    total = sum(flows)
    n = len(items)
    if n <= 1: return [0.5]
    usable = 1.0 - pad_norm * (n - 1)
    centers, cum = [], 0.0
    for f in flows:
        h = f / total * usable
        centers.append(cum + h / 2)
        cum += h + pad_norm
    return centers

def wrap_text(text, max_chars=55):
    words = text.split()
    lines, cur = [], ""
    for w in words:
        if len(cur) + len(w) + 1 <= max_chars:
            cur = (cur + " " + w).strip()
        else:
            if cur: lines.append(cur)
            cur = w
    if cur: lines.append(cur)
    return "<br>".join(lines)
```

```python
## MAIN FIGURE — standalone (no CSV needed)
# Edit the lists and dictionaries below to change labels or counts.

import plotly.graph_objects as go

# ── Config ────────────────────────────────────────────────────────────────────
SHOW_MIN    = 2
ALWAYS_SHOW = ["team", "teamwork", "collaborat"]

n_dp, n_dc, n_sr = len(vis_dp), len(vis_dc), len(vis_sr)

# DC flow totals (needed for proportional y positioning)
dc_totals = {dc: sum(v for (dp, d), v in layer1.items() if d == dc) for dc in vis_dc}
dc_y = flow_proportional_y(vis_dc, dc_totals)

# ── Colour palettes ───────────────────────────────────────────────────────────
dp_color_list = ["#4C72B0", "#8172B2", "#55A868", "#5A8F7B", "#DD8452"]
dp_colors     = dp_color_list[:n_dp]
dp_color_map  = dict(zip(vis_dp, dp_colors))

def dominant_dp_color(dc_name):
    best = max(
        ((dp, v) for (dp, d), v in layer1.items() if d == dc_name),
        key=lambda x: x[1], default=(None, 0)
    )
    return dp_color_map.get(best[0], "#CCCCCC")

dc_colors  = [dominant_dp_color(dc) for dc in vis_dc]
sr_palette = ["#C0392B","#E91E63","#F1C40F","#03A9F4","#795548","#607D8B","#00897B","#FF7043"]
sr_colors  = sr_palette[:n_sr]

# ── Build links ───────────────────────────────────────────────────────────────
left_nodes = vis_dp + vis_dc
left_index = {n: i for i, n in enumerate(left_nodes)}
left_colors_nodes = dp_colors + dc_colors
left_x = [0.01]*n_dp + [0.99]*n_dc
left_y = col_y(n_dp) + dc_y

ls, lt, lv = [], [], []
for (dp, dc), v in layer1.items():
    if dp in left_index and dc in left_index:
        ls.append(left_index[dc]); lt.append(left_index[dp]); lv.append(v)

left_link_colors = [
    hex_to_rgba(left_colors_nodes[s])
    if (v > SHOW_MIN or matches(left_nodes[t], ALWAYS_SHOW) or matches(left_nodes[s], ALWAYS_SHOW))
    else "rgba(0,0,0,0)"
    for s, t, v in zip(ls, lt, lv)
]

all_sr_right = vis_sr + hide_sr
right_nodes  = vis_dc + all_sr_right
right_index  = {n: i for i, n in enumerate(right_nodes)}
right_colors_nodes = dc_colors + sr_colors + ["rgba(0,0,0,0)"]*len(hide_sr)
right_x = [0.01]*n_dc + [0.99]*n_sr + [0.99]*len(hide_sr)
right_y  = dc_y + col_y(n_sr) + [0.01]*len(hide_sr)

rs, rt, rv, right_link_colors = [], [], [], []
for (dc, sr), v in layer2.items():
    if dc not in right_index or sr not in right_index: continue
    is_ghost = sr in hide_sr
    visible  = not is_ghost and (v > SHOW_MIN or matches(dc, ALWAYS_SHOW) or matches(sr, ALWAYS_SHOW))
    rs.append(right_index[dc]); rt.append(right_index[sr]); rv.append(v)
    right_link_colors.append(hex_to_rgba(right_colors_nodes[right_index[dc]]) if visible else "rgba(0,0,0,0)")

# ── Figure ────────────────────────────────────────────────────────────────────
fig = go.Figure()

fig.add_trace(go.Sankey(
    domain=dict(x=[0.04, 0.2], y=[0, 1]),
    arrangement="fixed",
    node=dict(pad=8, thickness=14, line=dict(color="white", width=0.5),
              label=[""]*len(left_nodes), color=left_colors_nodes,
              x=left_x, y=left_y,
              hovertemplate="<b>%{label}</b><br>%{value}<extra></extra>"),
    link=dict(source=ls, target=lt, value=lv, color=left_link_colors,
              hovertemplate="<b>%{source.label}</b>→<b>%{target.label}</b><br>%{value}<extra></extra>"),
))

fig.add_trace(go.Sankey(
    domain=dict(x=[0.8, 0.96], y=[0, 1]),
    arrangement="fixed",
    node=dict(pad=8, thickness=14, line=dict(color="rgba(0,0,0,0)", width=0),
              label=[""]*len(right_nodes), color=right_colors_nodes,
              x=right_x, y=right_y,
              hovertemplate="<b>%{label}</b><br>%{value}<extra></extra>"),
    link=dict(source=rs, target=rt, value=rv, color=right_link_colors,
              hovertemplate="<b>%{source.label}</b>→<b>%{target.label}</b><br>%{value}<extra></extra>"),
))

fig.update_layout(
    font=dict(family="Arial", size=13, color="#333333"),
    paper_bgcolor="white",
    height=500, width=1600,
    margin=dict(l=100, r=100, t=100, b=50),
)

# ── Annotations ───────────────────────────────────────────────────────────────
# Plotly Sankey: y=0 is TOP; paper annotations: y=0 is BOTTOM → use (1 - y_pos).

fig.add_annotation(x=0.5,   y=1.25, xref="paper", yref="paper",
    text="<b>Student Responses<br>Inductive Codes</b>", showarrow=False,
    font=dict(size=18, color="#444"))
fig.add_annotation(x=0, y=1.25,  xref="paper", yref="paper",
    text="<b>Deductively Mapped to<br>Design Principles</b>", showarrow=False,
    font=dict(size=18, color="#333"))
fig.add_annotation(x=1,  y=1.25,  xref="paper", yref="paper",
    text="<b>Deductively Mapped to<br>EVT/Self-Scaffolding</b>", showarrow=False,
    font=dict(size=18, color="#333"))

# DC labels: show student quote if one is mapped, otherwise render in white (hidden)
dc_y_nudge = {
    # "Debugging and seeing functional chat app was rewarding": +0.05,
    "Holistic understanding of db to app interactions":       -0.11,
}
for dc, y_pos in zip(vis_dc, dc_y):
    quote = dc_quotes.get(dc)
    if quote:
        nudge      = dc_y_nudge.get(dc, 0)
        label_text  = wrap_text(quote, max_chars=86)
        fig.add_annotation(x=0.5, y=1 - y_pos + nudge, xref="paper", yref="paper",
            text=label_text, showarrow=False,
            font=dict(size=18, color="#333333"), #, family='cursive'),
            xanchor="center", yanchor="middle", align="center")


for dp, y_pos in zip(vis_dp, col_y(n_dp)):
    fig.add_annotation(x=0.03, y=1 - y_pos, xref="paper", yref="paper",
        text=CATEGORY_LABELS[dp], showarrow=False,
        font=dict(size=18, color="#333333"),
        xanchor="right", yanchor="middle", align="right")

for sr, y_pos in zip(vis_sr, col_y(n_sr)):
    fig.add_annotation(x=0.97, y=1 - y_pos, xref="paper", yref="paper",
        text=sr, showarrow=False,
        font=dict(size=18, color="#333333"),
        xanchor="left", yanchor="middle", align="left")


fig.write_html("sankey_standalone.html")
print("✅  Saved: sankey_standalone.html")
fig.show()
```
